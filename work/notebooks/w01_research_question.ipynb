## 1. My lane and why

**Lane 2: Refresh / Content Opportunity Scoring.**
A website has too many pages to fix all of them, so the team needs to know which pages to check first. My idea is to start with pages that are performing badly but are still seen by enough people that fixing them is worth it. I chose this lane because the question is practical and the starter data has what I need (impressions, traffic trend, page age). 

```python
import pandas as pd
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")
print("Total pages:", len(df))
```

## 2. The question: decision, action, cost of a wrong call

**Decision:** Which pages should be checked first: the ones that people still see in Google, but whose traffic is falling.

**Who acts on it:** The content team member who reviews and updates the website's pages.

**Action:** The team reviews the top pages on the list, then updates, rewrites or keeps watching each one.

**Cost of a wrong call:** The team may spend time on pages that don't need fixing (for example, low traffic caused by season or low demand), and may miss pages that did need help.

```python
print(df["trend_direction"].value_counts(normalize=True).round(3))
```

Half of pages from dataset are loosing traffic. So they better need fixing first.

## 3. Quick look at the data (2-3 real numbers)



```python
seen = df["impressions_90d"] > df["impressions_90d"].median()
falling = df["trend_direction"] == "down"
n = (seen & falling).sum()
print(f"Seen a lot AND falling traffic: {n} ({n/len(df):.1%})")
```

The starter data has 30,000 pages, which is too many to review by hand. 8,906 pages (29.7%) are seen a lot (above the median impressions) but have falling traffic, so they are the natural first candidates for review.

## 4. Careful words: what I can and can't claim

**What I can say:** These numbers are observed in the starter data. They are directional and meant to support a decision about which pages to review first. "Seen a lot" means above the median impressions, which is my own cut-off, so a different cut-off could give different counts.

**What I can't say:** That falling traffic is caused by anything on the page, that fixing a page will recover its traffic, or that this predicts what Google will do. A page on the list may be falling for reasons like season or low demand, so it is a candidate for review, not a proven problem.

```python
falling = df["trend_direction"] == "down"

for q in [0.25, 0.50, 0.75]:
    cutoff = df["impressions_90d"].quantile(q)
    n = (falling & (df["impressions_90d"] > cutoff)).sum()
    print(f"Seen-a-lot cut-off at {int(q*100)}th percentile: {n} pages")
```

Depending on the cut-off for "seen a lot", the number of candidate pages ranges from 4,215 to 13,440.
